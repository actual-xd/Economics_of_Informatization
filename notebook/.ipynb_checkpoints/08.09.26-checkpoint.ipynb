Ноутбук содержит: критическую массу, прогноз пользователей, закон Мэткалфа, денежный поток, NPV, сценарии роста, безубыточный ARPU, маркетинговые инвестиции и стратегические выводы.

## Допущения и трактовка формул

1. Пользователи рассчитываются непрерывно. Это стандартное приближение для финансовой модели. В итогах число пользователей округляется до целого.
2. Пользовательский прирост начисляется в начале месяца на число пользователей предыдущего месяца. До достижения критической массы применяется `g_pre`, после — `g_post`.
3. Выручка месяца считается по числу пользователей на конец месяца: `R(N) = ARPU × N_end`.
4. Денежный поток месяца: `Прибыль = ARPU × N_end − поддержание − CAC × новые пользователи`. Первоначальные 5 000 пользователей считаются уже привлечёнными и не включаются в NPV.
5. Для пункта 1 используется буквальная формула задания: `ARPU × N ≥ затраты на поддержание + CAC`. Поэтому `N_кр = ceil((затраты + CAC) / ARPU)`.
6. Если трактовать CAC как стоимость всех новых пользователей каждый месяц, то при росте 3%: `ARPU − CAC × 3% = 15 − 500 × 0,03 = 0`. Такая строгая трактовка не даёт конечной критической массы для покрытия поддержки. Это ограничение показано отдельно, но основной расчёт следует формулировкам задания.

```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

plt.style.use('seaborn-v0_8-whitegrid')
pd.set_option('display.float_format', lambda value: f'{value:,.2f}')

# Исходные данные задания
MARKET_SIZE = 1_000_000
N0 = 5_000
PRE_GROWTH = 0.03
POST_GROWTH = 0.15
K = 0.005
MAINTENANCE = 2_000_000
CAC = 500
ARPU = 15
HORIZON = 36
DISCOUNT_RATE = 0.01

# Критическая масса по буквальному условию пункта 1
CRITICAL_MASS = math.ceil((MAINTENANCE + CAC) / ARPU)

print(f'Критическая масса: {CRITICAL_MASS:,} пользователей')
print(f'Доля от рынка: {CRITICAL_MASS / MARKET_SIZE:.2%}')
```

```python
def simulate(pre_growth, post_growth, horizon=HORIZON, arpu=ARPU, critical_mass=None):
    """Построить помесячную модель пользователей и денежных потоков."""
    # Для динамического анализа ARPU порог не округляем; основной порог передаём явно.
    threshold = (MAINTENANCE + CAC) / arpu if critical_mass is None else critical_mass
    users = float(N0)
    rows = []

    for month in range(1, horizon + 1):
        is_before_threshold = users < threshold and not np.isclose(users, threshold, rtol=0, atol=1e-8)
        growth = pre_growth if is_before_threshold else post_growth
        start_users = users
        new_users = min(start_users * growth, MARKET_SIZE - start_users)
        end_users = start_users + new_users
        revenue = arpu * end_users
        profit = revenue - MAINTENANCE - CAC * new_users
        discount_factor = 1 / (1 + DISCOUNT_RATE) ** month

        rows.append({
            'Месяц': month,
            'Пользователи_начало': start_users,
            'Новые_пользователи': new_users,
            'Пользователи_конец': end_users,
            'Темп_роста': growth,
            'Ценность_Мэткалфа': K * end_users ** 2,
            'Выручка': revenue,
            'Прибыль': profit,
            'Коэффициент_дисконтирования': discount_factor,
            'PV_прибыли': profit * discount_factor,
        })
        users = end_users

    result = pd.DataFrame(rows)
    reached = result.loc[result['Пользователи_конец'] >= threshold - 1e-8, 'Месяц']
    result.attrs['critical_mass'] = threshold
    result.attrs['month_reached'] = int(reached.iloc[0]) if not reached.empty else None
    return result


def npv(projection):
    return float(projection['PV_прибыли'].sum())


def money(value):
    return f'{value:,.0f} руб.'.replace(',', ' ')


def users(value):
    return f'{value:,.0f}'.replace(',', ' ')
```

## 1. Критическая масса и срок достижения

```python
critical_mass_recurring = math.ceil(MAINTENANCE / ARPU)
pre_growth_margin = ARPU - CAC * PRE_GROWTH
post_growth_margin = ARPU - CAC * POST_GROWTH

print('Критическая масса по условию задания:')
print(f'  N_кр = ceil(({MAINTENANCE:,} + {CAC}) / {ARPU}) = {CRITICAL_MASS:,} пользователей')
print(f'  Это {CRITICAL_MASS / MARKET_SIZE:.2%} потенциального рынка.')
print()
print('Справочно, без разового CAC:')
print(f'  N_кр = ceil({MAINTENANCE:,} / {ARPU}) = {critical_mass_recurring:,} пользователей')
print()
print('Проверка строгого ежемесячного учёта CAC × новые пользователи:')
print(f'  Маржа до критической массы = {ARPU} − {CAC} × {PRE_GROWTH:.0%} = {pre_growth_margin:.2f} руб. на пользователя')
print(f'  Маржа после критической массы = {ARPU} − {CAC} × {POST_GROWTH:.0%} = {post_growth_margin:.2f} руб. на пользователя')
print('  При такой трактовке переменные расходы до критической массы полностью съедают ARPU;')
print('  поэтому для расчёта срока ниже используется буквальное условие задания.')

reach_base = simulate(PRE_GROWTH, POST_GROWTH, horizon=400, critical_mass=CRITICAL_MASS)
month_reached = reach_base.attrs['month_reached']
users_at_reach = reach_base.loc[reach_base['Месяц'] == month_reached, 'Пользователи_конец'].iloc[0]

print()
print(f'Срок достижения критической массы: {month_reached} мес.')
print(f'Пользователей на конец месяца {month_reached}: {users(users_at_reach)}')
print(f'Критическая масса достигнута на {month_reached - HORIZON} мес. позже 36-месячного горизонта.' if month_reached > HORIZON else 'Критическая масса достигнута внутри 36-месячного горизонта.')
```

```python
# Траектория до достижения критической массы
fig, ax = plt.subplots(figsize=(11, 5.5))
ax.plot(reach_base['Месяц'], reach_base['Пользователи_конец'], color='#1769aa', linewidth=2.5, label='Пользователи')
ax.axhline(CRITICAL_MASS, color='#d62728', linestyle='--', linewidth=1.8, label=f'Критическая масса: {users(CRITICAL_MASS)}')
ax.axvline(month_reached, color='#2ca02c', linestyle=':', linewidth=1.8, label=f'Достижение: месяц {month_reached}')
ax.scatter([month_reached], [users_at_reach], color='#2ca02c', zorder=5)
ax.set_title('Динамика пользователей до достижения критической массы')
ax.set_xlabel('Месяц')
ax.set_ylabel('Число пользователей')
ax.legend()
plt.tight_layout()
plt.show()
```

## 2. Ценность сети, выручка и прибыль за 36 месяцев

```python
base = simulate(PRE_GROWTH, POST_GROWTH, horizon=HORIZON, critical_mass=CRITICAL_MASS)
base_display = base.copy()
base_display['Темп_роста'] = base_display['Темп_роста'].map(lambda value: f'{value:.0%}')
display(base_display)
print()
print(f'Пользователей на конец 36-го месяца: {users(base.iloc[-1]["Пользователи_конец"])}')
print(f'Ценность по Мэткалфу на конец 36-го месяца: {money(base.iloc[-1]["Ценность_Мэткалфа"])}')
print(f'Суммарная недисконтированная прибыль: {money(base["Прибыль"].sum())}')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 9))

axes[0, 0].plot(base['Месяц'], base['Пользователи_конец'], color='#1769aa', linewidth=2.5)
axes[0, 0].axhline(CRITICAL_MASS, color='#d62728', linestyle='--', label='Критическая масса')
axes[0, 0].set_title('Пользовательская база')
axes[0, 0].set_xlabel('Месяц')
axes[0, 0].set_ylabel('Пользователи')
axes[0, 0].legend()

axes[0, 1].plot(base['Месяц'], base['Ценность_Мэткалфа'] / 1_000_000, color='#7b2cbf', linewidth=2.5)
axes[0, 1].set_title('Ценность сети по закону Мэткалфа')
axes[0, 1].set_xlabel('Месяц')
axes[0, 1].set_ylabel('Млн руб.')

axes[1, 0].plot(base['Месяц'], base['Выручка'] / 1_000, color='#2ca02c', linewidth=2.5, label='Выручка')
axes[1, 0].axhline(MAINTENANCE / 1_000, color='#d62728', linestyle='--', label='Поддержание')
axes[1, 0].set_title('Выручка и фиксированные затраты')
axes[1, 0].set_xlabel('Месяц')
axes[1, 0].set_ylabel('Тысяч руб.')
axes[1, 0].legend()

colors = np.where(base['Прибыль'] >= 0, '#2ca02c', '#d62728')
axes[1, 1].bar(base['Месяц'], base['Прибыль'] / 1_000_000, color=colors)
axes[1, 1].axhline(0, color='black', linewidth=1)
axes[1, 1].set_title('Месячная прибыль / убыток')
axes[1, 1].set_xlabel('Месяц')
axes[1, 1].set_ylabel('Млн руб.')

fig.suptitle('Аналитика базового сценария, 36 месяцев', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

## 3. NPV и анализ сценариев роста

NPV рассчитан как сумма дисконтированных месячных прибылей: `NPV = Σ Прибыль_t / (1 + 0,01)^t`. Начальные 5 000 пользователей не являются расходом периода.

```python
scenarios = {
    'Пессимистичный': (0.02, 0.10),
    'Базовый': (0.03, 0.15),
    'Оптимистичный': (0.05, 0.20),
}

scenario_rows = []
for name, (pre, post) in scenarios.items():
    projection_36 = simulate(pre, post, horizon=HORIZON, critical_mass=CRITICAL_MASS)
    projection_long = simulate(pre, post, horizon=500, critical_mass=CRITICAL_MASS)
    reached_month = projection_long.attrs['month_reached']
    scenario_rows.append({
        'Сценарий': name,
        'До критической массы': f'{pre:.0%}',
        'После критической массы': f'{post:.0%}',
        'Месяц достижения': reached_month,
        'Пользователи на 36-м мес.': projection_36.iloc[-1]['Пользователи_конец'],
        'NPV за 36 мес.': npv(projection_36),
    })

scenario_table = pd.DataFrame(scenario_rows)
scenario_display = scenario_table.copy()
scenario_display['Пользователи на 36-м мес.'] = scenario_display['Пользователи на 36-м мес.'].map(users)
scenario_display['NPV за 36 мес.'] = scenario_display['NPV за 36 мес.'].map(money)
display(scenario_display)

base_npv = npv(base)
print(f'Базовый NPV за 36 месяцев: {money(base_npv)}')
print('Все сценарии дают отрицательный NPV в горизонте 36 месяцев: фиксированные затраты и CAC')
print('опережают выручку. Оптимистичный рост пока ухудшает NPV, потому что CAC = 500 руб.')
```

```python
fig, ax = plt.subplots(figsize=(10, 5.5))
scenario_colors = ['#d62728', '#1769aa', '#2ca02c']
bars = ax.bar(scenario_table['Сценарий'], scenario_table['NPV за 36 мес.'] / 1_000_000, color=scenario_colors)
ax.axhline(0, color='black', linewidth=1)
ax.set_title('NPV проекта при разных темпах роста')
ax.set_ylabel('NPV, млн руб.')
for bar, value in zip(bars, scenario_table['NPV за 36 мес.'] / 1_000_000):
    ax.text(bar.get_x() + bar.get_width() / 2, value - 1.2, f'{value:.1f}', ha='center', va='top', color='white', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 4. Минимальный ARPU для безубыточности

Есть два полезных ответа.

- **Фиксированная траектория пользователей**: темпы роста не меняются при изменении ARPU. Это консервативный ответ для финансового планирования NPV.
- **Динамическая модель**: ARPU меняет критическую массу, а достижение порога включает рост 15%. Такой ответ учитывает обратную связь модели и содержит скачок NPV в момент переключения режима.

```python
# 4.1. Безубыточный ARPU при неизменной базовой траектории пользователей
discount = base['Коэффициент_дисконтирования']
pv_users = (base['Пользователи_конец'] * discount).sum()
pv_maintenance = MAINTENANCE * discount.sum()
pv_cac = (CAC * base['Новые_пользователи'] * discount).sum()
break_even_arpu_fixed = (pv_maintenance + pv_cac) / pv_users

print(f'Безубыточный ARPU при фиксированной траектории: {break_even_arpu_fixed:.2f} руб./пользователь/мес.')
print(f'Рост относительно текущего ARPU: {(break_even_arpu_fixed / ARPU - 1):.1%}')

# 4.2. Динамический порог: ARPU, при котором порог достигается в конце 24-го месяца.
# На следующем месяце модель переходит к темпу 15%, и NPV становится неотрицательным.
users_month_24_pre = N0 * (1 + PRE_GROWTH) ** 24
break_even_arpu_dynamic = (MAINTENANCE + CAC) / users_month_24_pre
dynamic_projection = simulate(PRE_GROWTH, POST_GROWTH, horizon=HORIZON, arpu=break_even_arpu_dynamic)
dynamic_npv = npv(dynamic_projection)

print(f'Динамический пороговый ARPU: {break_even_arpu_dynamic:.2f} руб./пользователь/мес.')
print(f'NPV при динамическом пороге: {money(dynamic_npv)}')
print(f'Переключение на 15%: после месяца {dynamic_projection.attrs["month_reached"]}')
print('Для консервативного бизнес-плана рекомендуется использовать фиксированный ответ:')
print(f'{break_even_arpu_fixed:.2f} руб./пользователь/мес.')
```

```python
arpu_grid = np.linspace(10, 260, 251)
dynamic_npv_grid = []
fixed_npv_grid = []

for arpu_value in arpu_grid:
    fixed_npv_grid.append(arpu_value * pv_users - pv_maintenance - pv_cac)
    dynamic_npv_grid.append(npv(simulate(PRE_GROWTH, POST_GROWTH, horizon=HORIZON, arpu=arpu_value)))

fig, ax = plt.subplots(figsize=(11, 5.5))
ax.plot(arpu_grid, np.array(fixed_npv_grid) / 1_000_000, label='Фиксированная траектория', linewidth=2.3)
ax.plot(arpu_grid, np.array(dynamic_npv_grid) / 1_000_000, label='Динамическая траектория', linewidth=2.0, alpha=0.85)
ax.axhline(0, color='black', linewidth=1)
ax.axvline(break_even_arpu_fixed, color='#1769aa', linestyle='--', label=f'Фиксированный порог: {break_even_arpu_fixed:.2f}')
ax.axvline(break_even_arpu_dynamic, color='#2ca02c', linestyle=':', label=f'Динамический порог: {break_even_arpu_dynamic:.2f}')
ax.set_title('Чувствительность NPV к ARPU')
ax.set_xlabel('ARPU, руб./пользователь/мес.')
ax.set_ylabel('NPV, млн руб.')
ax.legend()
plt.tight_layout()
plt.show()
```

## 5. Маркетинговые инвестиции для ускорения на 6 месяцев

Базовый срок — 112 месяцев. Целевой срок — 106 месяцев. Самый экономичный вариант — привлечь дополнительные пользователи в начале проекта: они успеют вырасти вместе с сетью. Для сравнения приведён вариант закупки пользователей непосредственно в 106-м месяце.

```python
target_month = month_reached - 6

# Вариант A: дополнительные пользователи в начале проекта
required_initial_users = CRITICAL_MASS / (1 + PRE_GROWTH) ** target_month
additional_initial_users = math.ceil(required_initial_users - N0)
initial_investment = additional_initial_users * CAC

# Вариант B: докупить пользователей в целевом месяце
base_users_at_target = N0 * (1 + PRE_GROWTH) ** target_month
additional_target_users = math.ceil(CRITICAL_MASS - base_users_at_target)
target_month_investment = additional_target_users * CAC

marketing_table = pd.DataFrame({
    'Вариант': ['Дополнительное привлечение в начале', 'Докупить в 106-м месяце'],
    'Дополнительные пользователи': [additional_initial_users, additional_target_users],
    'Инвестиции': [initial_investment, target_month_investment],
    'Результат': ['Критическая масса к 106-му месяцу', 'Критическая масса к 106-му месяцу'],
})
marketing_display = marketing_table.copy()
marketing_display['Дополнительные пользователи'] = marketing_display['Дополнительные пользователи'].map(users)
marketing_display['Инвестиции'] = marketing_display['Инвестиции'].map(money)
display(marketing_display)

print(f'Рекомендуемый минимальный бюджет: {money(initial_investment)}')
print(f'Он требует привлечь {users(additional_initial_users)} дополнительных пользователей в начале проекта.')
print(f'Если ждать до месяца {target_month}, бюджет возрастает до {money(target_month_investment)}')
```

## 6. Стратегические выводы и рекомендации

1. **Критическая масса пока недостижима в горизонте 36 месяцев.** При базовом росте сеть достигает порога только в 112-м месяце и к концу третьего года имеет около 14,5 тыс. пользователей.
2. **Главный финансовый риск — низкая выручка относительно фиксированных затрат.** Базовый NPV за 36 месяцев — около −60,1 млн руб.
3. **Ускорение роста само по себе не гарантирует улучшение NPV.** При CAC 500 руб. и ARPU 15 руб. дополнительные пользователи на раннем этапе увеличивают затраты быстрее выручки. Поэтому оптимистичный сценарий даёт ещё более отрицательный NPV.
4. **Нужно повысить ARPU или снизить затраты.** При фиксированной траектории безубыточный ARPU — около 242,55 руб. в месяц. Альтернативы: подписка, платные функции, реклама, партнёрские размещения, снижение поддержки и CAC.
5. **Маркетинг следует запускать ступенчато.** Рекомендуется тестировать каналы на небольших когортах, считать CAC по активированному пользователю и масштабировать только каналы с доказанным удержанием.
6. **Для ускорения на шесть месяцев выгоднее раннее привлечение.** Ориентир — 812 дополнительных пользователей в начале и бюджет около 406 тыс. руб. Закупка того же объёма в 106-м месяце потребует около 9,31 млн руб.
7. **Критическую массу нужно уточнить в финансовой модели.** При строгом ежемесячном списании CAC переменная маржа при росте 3% равна нулю. Поэтому бизнес-план должен разделить разовые расходы привлечения, маркетинговый бюджет и постоянные расходы сети.

```python
summary = pd.DataFrame({
    'Показатель': [
        'Критическая масса',
        'Срок достижения',
        'Пользователи на 36-м месяце',
        'NPV за 36 месяцев',
        'Безубыточный ARPU, фиксированная траектория',
        'Минимальный бюджет ускорения на 6 месяцев',
    ],
    'Значение': [
        users(CRITICAL_MASS),
        f'{month_reached} мес.',
        users(base.iloc[-1]['Пользователи_конец']),
        money(base_npv),
        f'{break_even_arpu_fixed:.2f} руб./мес.',
        money(initial_investment),
    ]
})
display(summary)
```
