# Лабораторная работа №3. Расчёт критической массы и ценности социальной сети

**Компания «ГородСоцСеть»** запускает социальную сеть для города с населением 1 млн человек. Нужно оценить экономическую ценность сети и определить момент достижения критической массы.

Ноутбук повторяет структуру задания: расчёт критической массы, оценка ценности сети (закон Мэткалфа), анализ эффективности (NPV, сценарии, безубыточный ARPU) и стратегические выводы.


## Исходные данные

| Параметр | Обозначение | Значение |
|---|---|---|
| Общий потенциал рынка | M | 1 000 000 чел. |
| Начальное число пользователей | N₀ | 5 000 чел. |
| Темп прироста до критической массы | g_pre | 3% в месяц |
| Темп прироста после критической массы | g_post | 15% в месяц |
| Коэффициент ценности сети | k | 0,005 руб./пользователь² |
| Затраты на поддержание сети | Затраты | 2 000 000 руб./мес. |
| Стоимость привлечения пользователя | CAC | 500 руб. |
| Средний доход с пользователя | ARPU | 15 руб./мес. |
| Горизонт | T | 36 мес. |
| Ставка дисконтирования | r | 1% в месяц |

**Основные формулы**

- Число пользователей: `N_t = N_{t-1} × (1 + g)`
- Ценность сети (закон Мэткалфа): `V(N) = k × N²`
- Доход: `R(N) = ARPU × N`
- Прибыль: `П(N) = R(N) − Затраты − CAC × (новые пользователи)`
- NPV: `NPV = Σ П_t / (1 + r)^t`


## Допущения

1. Число пользователей считается непрерывным (стандартное приближение финансовой модели), в отчёте округляется до целого.
2. Прирост начисляется в начале месяца на базу предыдущего месяца. До критической массы применяется `g_pre = 3%`, после — `g_post = 15%`.
3. Выручка месяца считается по числу пользователей на конец месяца.
4. Начальные 5 000 пользователей считаются уже привлечёнными и не создают разового оттока CAC.
5. Критическая масса определяется по формуле задания: `ARPU × N ≥ Затраты + CAC`.


```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

plt.style.use('seaborn-v0_8-whitegrid')
pd.set_option('display.float_format', lambda value: f'{value:,.2f}')

# Исходные данные
M = 1_000_000            # потенциал рынка, чел.
N0 = 5_000               # начальное число пользователей, чел.
G_PRE = 0.03             # темп прироста до критической массы
G_POST = 0.15            # темп прироста после критической массы
K = 0.005                # коэффициент ценности сети, руб./чел.^2
MAINTENANCE = 2_000_000  # затраты на поддержание, руб./мес.
CAC = 500                # стоимость привлечения пользователя, руб.
ARPU = 15                # доход с пользователя, руб./мес.
HORIZON = 36             # горизонт, мес.
RATE = 0.01              # ставка дисконтирования, 1% в месяц

# Критическая масса по формуле задания: ARPU * N >= поддержание + CAC
CRITICAL_MASS = math.ceil((MAINTENANCE + CAC) / ARPU)


def fmt_money(value):
    return f'{value:,.0f} руб.'.replace(',', ' ')


def fmt_users(value):
    return f'{value:,.0f}'.replace(',', ' ')


def simulate(g_pre, g_post, horizon=HORIZON, arpu=ARPU, threshold=CRITICAL_MASS):
    """Помесячная модель: пользователи, доход, прибыль и дисконтированный поток."""
    users = float(N0)
    rows = []
    for month in range(1, horizon + 1):
        growth = g_pre if users < threshold else g_post
        new_users = min(users * growth, M - users)
        end_users = users + new_users
        revenue = arpu * end_users
        profit = revenue - MAINTENANCE - CAC * new_users
        discount = 1 / (1 + RATE) ** month
        rows.append({
            'Месяц': month,
            'Пользователи': end_users,
            'Новые': new_users,
            'Темп': growth,
            'Ценность V=kN²': K * end_users ** 2,
            'Доход R=ARPU×N': revenue,
            'Прибыль П': profit,
            'Дисконт': discount,
            'PV прибыли': profit * discount,
        })
        users = end_users

    frame = pd.DataFrame(rows)
    reached = [int(m) for m, n in zip(frame['Месяц'], frame['Пользователи']) if n >= threshold]
    frame.attrs['month_reached'] = reached[0] if reached else None
    return frame


def npv(frame):
    return float(frame['PV прибыли'].sum())


print('Модель готова. Критическая масса по условию:', fmt_users(CRITICAL_MASS), 'пользователей')
```

## Часть 1. Расчёт критической массы

**Критическая масса** — минимальное число пользователей, при котором сеть самоподдерживается:

`ARPU × N ≥ Затраты на поддержание + CAC`

Поэтому `N_кр = (Затраты + CAC) / ARPU`.


```python
recurring_mass = math.ceil(MAINTENANCE / ARPU)
margin_pre = ARPU - CAC * G_PRE
margin_post = ARPU - CAC * G_POST

print('Критическая масса по формуле задания:')
print(f'  N_кр = ({MAINTENANCE:,} + {CAC}) / {ARPU} = {(MAINTENANCE + CAC) / ARPU:,.2f}')
print(f'  N_кр = {fmt_users(CRITICAL_MASS)} пользователей')
print(f'  Это {CRITICAL_MASS / M:.2%} потенциального рынка.')
print()
print('Справочно, если CAC учитывать разово, а поддержание — ежемесячно:')
print(f'  N_кр = {MAINTENANCE:,} / {ARPU} = {fmt_users(recurring_mass)} пользователей')
print()
print('Проверка переменной маржи на одного пользователя:')
print(f'  до критической массы:  ARPU − CAC × g_pre  = {ARPU} − {CAC} × {G_PRE:.0%} = {margin_pre:.2f} руб.')
print(f'  после критической массы: ARPU − CAC × g_post = {ARPU} − {CAC} × {G_POST:.0%} = {margin_post:.2f} руб.')
print()
print('Вывод: при росте 3% и CAC 500 руб. вся выручка уходит на привлечение,')
print('поэтому критическую массу определяем по буквальной формуле задания.')
```

### Время достижения критической массы

При постоянном темпе прироста `g` пользователи растут экспоненциально: `N_t = N₀ × (1 + g)^t`.

Момент достижения находится аналитически: `t = ln(N_кр / N₀) / ln(1 + g)`.


```python
path = simulate(G_PRE, G_POST, horizon=400)
month_reached = path.attrs['month_reached']
analytic_months = math.log(CRITICAL_MASS / N0) / math.log(1 + G_PRE)
users_at_reach = path.loc[path['Месяц'] == month_reached, 'Пользователи'].iloc[0]

print(f't = ln({CRITICAL_MASS / N0:.2f}) / ln(1.03) = {analytic_months:.2f} мес.')
print(f'Критическая масса достигается на {month_reached}-м месяце прироста по 3%.')
print(f'Пользователей в этот момент: {fmt_users(users_at_reach)}')
print(f'Это на {month_reached - HORIZON} мес. позже 36-месячного горизонта.')

fig, ax = plt.subplots(figsize=(11, 5.5))
ax.plot(path['Месяц'], path['Пользователи'], color='#1769aa', linewidth=2.5, label='Пользователи')
ax.axhline(CRITICAL_MASS, color='#d62728', linestyle='--', linewidth=1.8,
           label=f'Критическая масса: {fmt_users(CRITICAL_MASS)}')
ax.axvline(month_reached, color='#2ca02c', linestyle=':', linewidth=1.8,
           label=f'Достижение: {month_reached} мес.')
ax.scatter([month_reached], [users_at_reach], color='#2ca02c', zorder=5)
ax.axvline(HORIZON, color='gray', linestyle='-', linewidth=1.2, label='Горизонт 36 мес.')
ax.set_title('Динамика пользователей при росте 3% в месяц')
ax.set_xlabel('Месяц')
ax.set_ylabel('Число пользователей')
ax.legend()
plt.tight_layout()
plt.show()
```

## Часть 2. Оценка ценности сети

Для каждого месяца считаются:

- ценность сети по закону Мэткалфа `V(N) = k × N²` — сеть ценна связями, поэтому ценность растёт как квадрат числа участников;
- доход `R(N) = ARPU × N`;
- прибыль `П(N) = R(N) − Затраты − CAC × новые пользователи`.


```python
base = simulate(G_PRE, G_POST, horizon=HORIZON)
base_view = base.copy()
base_view['Темп'] = base_view['Темп'].map(lambda value: f'{value:.0%}')
display(base_view)

last = base.iloc[-1]
print(f'Пользователей на конец 36-го месяца: {fmt_users(last["Пользователи"])}')
print(f'Ценность сети по Мэткалфу: {fmt_money(last["Ценность V=kN²"])}')
print(f'Доход за 36-й месяц: {fmt_money(last["Доход R=ARPU×N"])}')
print(f'Суммарный недисконтированный убыток за 36 мес.: {fmt_money(base["Прибыль П"].sum())}')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 9))

axes[0, 0].plot(base['Месяц'], base['Пользователи'], color='#1769aa', linewidth=2.5)
axes[0, 0].axhline(CRITICAL_MASS, color='#d62728', linestyle='--', label='Критическая масса')
axes[0, 0].set_title('Пользовательская база')
axes[0, 0].set_xlabel('Месяц')
axes[0, 0].set_ylabel('Пользователи')
axes[0, 0].legend()

axes[0, 1].plot(base['Месяц'], base['Ценность V=kN²'] / 1_000_000, color='#7b2cbf', linewidth=2.5)
axes[0, 1].set_title('Ценность сети по закону Мэткалфа')
axes[0, 1].set_xlabel('Месяц')
axes[0, 1].set_ylabel('Млн руб.')

axes[1, 0].plot(base['Месяц'], base['Доход R=ARPU×N'] / 1_000, color='#2ca02c', linewidth=2.5, label='Доход')
axes[1, 0].axhline(MAINTENANCE / 1_000, color='#d62728', linestyle='--', label='Поддержание')
axes[1, 0].set_title('Доход и фиксированные затраты')
axes[1, 0].set_xlabel('Месяц')
axes[1, 0].set_ylabel('Тыс. руб.')
axes[1, 0].legend()

colors = np.where(base['Прибыль П'] >= 0, '#2ca02c', '#d62728')
axes[1, 1].bar(base['Месяц'], base['Прибыль П'] / 1_000_000, color=colors)
axes[1, 1].axhline(0, color='black', linewidth=1)
axes[1, 1].set_title('Месячная прибыль / убыток')
axes[1, 1].set_xlabel('Месяц')
axes[1, 1].set_ylabel('Млн руб.')

fig.suptitle('Базовый сценарий: сеть, ценность и денежный поток за 36 месяцев', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

## Часть 3. Анализ эффективности

### 3.1. NPV за 36 месяцев

`NPV = Σ П_t / (1 + r)^t`, где `r = 1%` в месяц.


```python
base_npv = npv(base)

print('NPV базового сценария за 36 месяцев:')
print(f'  NPV = Σ П_t / (1,01)^t = {fmt_money(base_npv)}')
print()
print('При росте 3% переменная маржа равна нулю (ARPU = CAC × 3% = 15 руб.),')
print('поэтому ежемесячная прибыль равна минус фиксированным затратам:')
print(f'  П = − {fmt_money(MAINTENANCE)} каждый месяц.')
```

### 3.2. Сценарии темпов роста

- пессимистичный: 2% до критической массы, 10% после;
- базовый: 3% / 15%;
- оптимистичный: 5% / 20%.


```python
scenarios = {
    'Пессимистичный': (0.02, 0.10),
    'Базовый': (0.03, 0.15),
    'Оптимистичный': (0.05, 0.20),
}

scenario_rows = []
for name, (pre, post) in scenarios.items():
    projection = simulate(pre, post, horizon=HORIZON)
    scenario_rows.append({
        'Сценарий': name,
        'До критической массы': f'{pre:.0%}',
        'После критической массы': f'{post:.0%}',
        'Пользователи на 36-м мес.': projection.iloc[-1]['Пользователи'],
        'NPV за 36 мес.': npv(projection),
    })

scenario_table = pd.DataFrame(scenario_rows)
scenario_view = scenario_table.copy()
scenario_view['Пользователи на 36-м мес.'] = scenario_view['Пользователи на 36-м мес.'].map(fmt_users)
scenario_view['NPV за 36 мес.'] = scenario_view['NPV за 36 мес.'].map(fmt_money)
display(scenario_view)
```

```python
fig, ax = plt.subplots(figsize=(10, 5.5))
scenario_colors = ['#d62728', '#1769aa', '#2ca02c']
bars = ax.bar(scenario_table['Сценарий'], scenario_table['NPV за 36 мес.'] / 1_000_000, color=scenario_colors)
ax.axhline(0, color='black', linewidth=1)
ax.set_title('NPV проекта при разных темпах роста')
ax.set_ylabel('NPV, млн руб.')
for bar, value in zip(bars, scenario_table['NPV за 36 мес.'] / 1_000_000):
    offset = 0.6 if value >= 0 else -0.6
    va = 'bottom' if value >= 0 else 'top'
    ax.text(bar.get_x() + bar.get_width() / 2, value + offset, f'{value:.1f}', ha='center', va=va)
plt.tight_layout()
plt.show()
```

### 3.3. Минимальный ARPU для безубыточности

Проект безубыточен, когда `NPV = 0`. При неизменной (базовой) траектории пользователей:

`ARPU × Σ N_t/(1+r)^t = Σ (Затраты + CAC × новые_t)/(1+r)^t`


```python
discount = base['Дисконт']
pv_users = (base['Пользователи'] * discount).sum()
pv_cost = (MAINTENANCE * discount).sum() + (CAC * base['Новые'] * discount).sum()
break_even_arpu = pv_cost / pv_users

print(f'Дисконтированная база пользователей: {fmt_users(pv_users)}')
print(f'Дисконтированные затраты:            {fmt_money(pv_cost)}')
print(f'Минимальный безубыточный ARPU:       {break_even_arpu:.2f} руб./пользователь/мес.')
print(f'Это в {break_even_arpu / ARPU:.1f} раза больше текущего ARPU = {ARPU} руб.')
print()
print('Причина: при ARPU = 15 руб. и росте 3% переменная маржа равна нулю,')
print('фиксированные 2 млн руб./мес. не покрываются, NPV остаётся отрицательным.')
```

```python
arpu_grid = np.linspace(0, 300, 301)
npv_grid = []
for arpu_value in arpu_grid:
    projection = simulate(G_PRE, G_POST, horizon=HORIZON, arpu=arpu_value)
    npv_grid.append(npv(projection))

fig, ax = plt.subplots(figsize=(11, 5.5))
ax.plot(arpu_grid, np.array(npv_grid) / 1_000_000, color='#1769aa', linewidth=2.5)
ax.axhline(0, color='black', linewidth=1)
ax.axvline(break_even_arpu, color='#d62728', linestyle='--',
           label=f'Безубыточный ARPU: {break_even_arpu:.2f} руб.')
ax.axvline(ARPU, color='#2ca02c', linestyle=':', label=f'Текущий ARPU: {ARPU} руб.')
ax.set_title('Чувствительность NPV к ARPU')
ax.set_xlabel('ARPU, руб./пользователь/мес.')
ax.set_ylabel('NPV, млн руб.')
ax.legend()
plt.tight_layout()
plt.show()
```

## Часть 4. Стратегические выводы

### 4.1. Маркетинговые инвестиции для ускорения на 6 месяцев

Цель — достичь критической массы за `t_target = t_base − 6 = 106` месяцев. Нужно привлечь дополнительных пользователей **в начале проекта**, чтобы они выросли вместе с сетью:

`N₀' = N_кр / (1 + g_pre)^t_target`


```python
target_month = month_reached - 6
required_start = CRITICAL_MASS / (1 + G_PRE) ** target_month
extra_users = math.ceil(required_start - N0)
budget_early = extra_users * CAC

base_at_target = N0 * (1 + G_PRE) ** target_month
need_at_target = math.ceil(CRITICAL_MASS - base_at_target)
budget_late = need_at_target * CAC

marketing_table = pd.DataFrame({
    'Вариант': ['Привлечь в начале проекта', f'Докупить в {target_month}-м месяце'],
    'Дополнительные пользователи': [extra_users, need_at_target],
    'Инвестиции': [budget_early, budget_late],
})
marketing_view = marketing_table.copy()
marketing_view['Дополнительные пользователи'] = marketing_view['Дополнительные пользователи'].map(fmt_users)
marketing_view['Инвестиции'] = marketing_view['Инвестиции'].map(fmt_money)
display(marketing_view)

print(f'Целевой месяц: {target_month} (на 6 мес. раньше {month_reached}-го).')
print(f'Минимальный бюджет: {fmt_money(budget_early)} — привлечь {fmt_users(extra_users)} чел. в начале.')
print(f'Если ждать до {target_month}-го месяца: {fmt_money(budget_late)} — в {budget_late / budget_early:.1f} раза дороже.')
```

### 4.2. Рекомендации для «ГородСоцСеть»

1. **Критическая масса недостижима в горизонте 36 месяцев.** При росте 3% порог 133 367 пользователей достигается только на 112-м месяце; к концу третьего года база — около 14,5 тыс. человек (1,4% рынка).
2. **Главный риск — экономика юнита.** При ARPU 15 руб. и CAC 500 руб. привлечение при росте 3% съедает всю выручку (15 = 500 × 3%). Бизнес не масштабируется за счёт роста сам по себе.
3. **Ускорение роста ухудшает NPV при текущей экономике.** Чем быстрее рост, тем больше новых пользователей и расходов на CAC, поэтому оптимистичный сценарий даёт NPV хуже базового.
4. **Ключевая задача — поднять ARPU или снизить CAC/затраты.** Безубыточный ARPU при текущей траектории — около 242,6 руб./мес. (в 16 раз выше). Варианты: подписка, платные функции, реклама, партнёрские размещения, снижение CAC через виральные каналы.
5. **Маркетинг — ступенчато.** Запускать на небольших когортах, считать CAC по активированному и удержанному пользователю, масштабировать только каналы с доказанным удержанием.
6. **Ускорение на 6 месяцев дешевле делать ранним привлечением.** Ориентир — 812 дополнительных пользователей в начале и бюджет 406 тыс. руб.; докупка того же объёма в 106-м месяце стоила бы около 9,3 млн руб.
7. **Разделить затраты в финансовой модели.** Разовый CAC, маркетинговый бюджет и постоянные затраты сети нужно учитывать отдельно — иначе критическая масса и точка безубыточности искажаются.


```python
summary = pd.DataFrame({
    'Показатель': [
        'Критическая масса',
        'Срок достижения критической массы',
        'Пользователи на 36-м месяце',
        'NPV за 36 месяцев (базовый)',
        'NPV, пессимистичный сценарий',
        'NPV, оптимистичный сценарий',
        'Безубыточный ARPU',
        'Минимальный бюджет ускорения на 6 месяцев',
    ],
    'Значение': [
        fmt_users(CRITICAL_MASS),
        f'{month_reached} мес.',
        fmt_users(base.iloc[-1]['Пользователи']),
        fmt_money(base_npv),
        fmt_money(scenario_table.loc[scenario_table['Сценарий'] == 'Пессимистичный', 'NPV за 36 мес.'].iloc[0]),
        fmt_money(scenario_table.loc[scenario_table['Сценарий'] == 'Оптимистичный', 'NPV за 36 мес.'].iloc[0]),
        f'{break_even_arpu:.2f} руб./мес.',
        fmt_money(budget_early),
    ],
})
display(summary)
```

## Вывод

Критическая масса сети по условию задания — **133 367 пользователей** (13,3% рынка). При базовом росте 3% в месяц она достигается только на **112-м месяце**, то есть за пределами 36-месячного горизонта.

Ценность сети по закону Мэткалфа к концу 36-го месяца — около **1,05 млн руб.**, но достигнута она на базе всего около 14,5 тыс. пользователей. Доход растёт линейно (ARPU × N) и остаётся ниже фиксированных затрат 2 млн руб./мес.

NPV базового сценария за 36 месяцев — около **−60,2 млн руб.**; пессимистичный и оптимистичный сценарии также отрицательны. Рост сам по себе не спасает проект: при CAC 500 руб. и ARPU 15 руб. переменная маржа равна нулю, а ускоренный рост увеличивает расходы на привлечение.

Проект становится безубыточным только при ARPU около **242,6 руб./мес.** (в 16 раз выше текущего) либо при существенном снижении затрат и CAC. Ускорение достижения критической массы на 6 месяцев требует **406 тыс. руб.** дополнительных маркетинговых инвестиций при раннем привлечении пользователей.
